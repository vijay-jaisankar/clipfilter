```python
# Create a Matplotlib figure and axes
fig, ax = plt.subplots(1, 1, figsize=(12, 8))

# Plot the world map as the base layer
world.plot(ax=ax, color="lightgray", edgecolor=None)

# True Ground Truth
gdf.plot(
    ax=ax,
    marker="*",
    color="yellow",
    markersize=30,
    label="Actual Ground Truth Coordinates",
)

# GeoCLIP Regular
gdf2.plot(
    ax=ax,
    marker="*",
    color="red",
    markersize=5,
    label="GeoCLIP Predictions for Unaltered images",
)

# GeoCLIP Filter
gdf3.plot(
    ax=ax,
    marker="o",
    color="blue",
    markersize=5,
    label="GeoCLIP Predictions for Images with filter applied",
)

# OSMCLIP Regular
gdf4.plot(
    ax=ax,
    marker="*",
    color="green",
    markersize=5,
    label="OSMCLIP Predictions for Unaltered images",
)

# OSMCLIP Filter
gdf5.plot(
    ax=ax,
    marker="o",
    color="purple",
    markersize=5,
    label="OSMCLIP Predictions for Images with filter applied",
)


# Add a title and legend for clarity
ax.set_title("Effects of Synthetic filters on GeoCLIP and OSMCLIP model")
ax.legend(loc="lower left")

# Show the plot
plt.show()
```
